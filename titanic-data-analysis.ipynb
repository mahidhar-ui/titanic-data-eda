```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
iris=sns.load_dataset("iris")
iris
```

```python
iris.species.value_counts()
```

```python
iris.plot(kind="scatter",x="sepal_length",y="sepal_width")
```

```python
colors={'setosa':'red','versicolor':'blue','virginica':'green'}
for species in iris ['species'].unique():
  subset=iris[iris['species']==species]
  plt.scatter(subset['sepal_length'],subset['sepal_width'],
              color=colors[species],label=species)
```

```python
sns.scatterplot(data=iris,x="sepal_length",y="sepal_width",hue="species")
plt.legend(title="species")
plt.grid(True)
plt.show()
```

```python
import plotly.express as px
import plotly.graph_objects as go
fig=px.scatter(iris,x="sepal_length",y="sepal_width",color="species")
fig.show()
```

```python
sns.pairplot (iris,hue='species')
plt.suptitle("pairplot of iris dataset")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=iris,x="sepal_length",hue='species',bins=20)
```

```python
sns.boxplot(data=iris,x="species",y="sepal_length")
```

```python
virginica=iris[iris["species"]=="virginica"]
virginica
q1=virginica["sepal_length"].quantile(0.25)
q3=virginica["sepal_length"].quantile(0.75)
iqr=q3-q1
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr
outliers=virginica[(virginica["sepal_length"]<lower_bound) |
                   (virginica["sepal_length"]>upper_bound)]
outliers
```

```python
sns.lineplot(data=iris,x="sepal_length",y="sepal_width",hue="species")
plt.show()
```

```python
corr_petal=iris['petal_length'].corr(iris['petal_width'])
corr_sepal=iris['sepal_length'].corr(iris['sepal_width'])
corr_petal,corr_sepal
```

```python
corr_mahi=iris[['sepal_length','sepal_width','petal_length','petal_width']].corr()
corr_mahi
```

```python
sns.heatmap(corr_mahi,annot=True,cmap='coolwarm')
```
